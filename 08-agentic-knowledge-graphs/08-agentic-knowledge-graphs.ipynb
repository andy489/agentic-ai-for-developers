### 1.3 Entities, Relationships, Ontologies, and Schemas

#### The Four Building Blocks of a Knowledge Graph

| Building Block | Role |
|---|---|
| **Entities** | The things in our data |
| **Relationships** | How those things connect |
| **Ontologies** | Rules about allowed relationships |
| **Schemas** | Rules that ensure consistency |

#### Entities: The Things Orion Cares About

Orion's knowledge graph has four entity types:
- **Customers** — people who report problems
- **Products** — the software or hardware involved
- **Issues** — the problems experienced
- **Resolutions** — the fixes applied

#### Relationships: How Entities Connect

Relationships are directed, typed edges between nodes:

- `reported` — links a customer to an issue
- `affects` — links an issue to a product
- `resolved_by` — links an issue to a resolution

#### Ontologies Define Allowed Connections

Ontologies encode **domain rules** about which relationships are valid:

- Customer `reports` Issues ✓
- Issue `affects` Products ✓
- A Product **cannot** report an Issue ✗ (forbidden)

Ontologies act as the **grammar** of the graph — they prevent nonsensical connections.

#### Schemas Keep the Graph Consistent

Schemas define **how data must be structured** and enforce required fields:

```
Issue:      id, description, severity
Customer:   customer_id, name
Product:    product_id, version
Resolution: resolution_id, steps, status
```

Schemas ensure each node and edge uses the correct types and directions.

### 1.4 How Agents Traverse Graphs

#### Nodes, Edges, and Triplets

Every fact in a knowledge graph is a **triplet**:

```
(Subject) --[Predicate]--> (Object)

Example:
(Customer A) --[reported]--> (Battery Drain Issue)
(Battery Drain Issue) --[affects]--> (Product X)
(Battery Drain Issue) --[resolved_by]--> (Patch 1.2)
```

#### Traversal: Following the Edges

An agent traverses the graph by following edges from node to node. This enables **multi-hop reasoning** — chaining multiple relationships to reach a conclusion:

```
Customer A
  → reported → Battery Drain Issue
    → affects → Product X
      → has_update → Latest Update
        → includes → Patch 1.2
```

Each hop adds context. The path itself is the **evidence trail**.

#### Nodes and Attributes

Nodes carry **attributes** — key-value properties that describe the entity in more detail:

```
Issue node:
  id: "ISS-042"
  title: "Battery drain"
  severity: "high"
  status: "open"
```

### 1.5 How Orion Searches Today

Before adding a knowledge graph, Orion relies on two retrieval strategies:

#### Document Retrieval

- Searches for **exact keyword matches** in stored tickets and documents
- Strength: precise matches for known terms
- Weakness: **too literal** — misses semantic variations and synonyms

#### Vector Retrieval

- Encodes text as **semantic embeddings** and finds similar vectors
- Strength: captures meaning even when wording differs
- Weakness: **can't explain relationships** — no causal or structural understanding

#### The Missing Layer: Graph Reasoning

Document and vector retrieval answer *what* but not *why*. The graph adds the missing layer:

```
Product X → Overheating Issue → Latest Update → Patch 1.2
```

This chain shows *cause and effect* — something neither document nor vector search can surface.

#### How Each Method Helps Orion

| Method | Strength | Weakness |
|---|---|---|
| **Document Retrieval** | Exact keyword matches | Misses meaning, too literal |
| **Vector Retrieval** | Semantic similarity | Lacks structure and causality |
| **Graph Reasoning** | Structure and causal paths | Requires structured data |

#### Completing Orion's Intelligence Stack

The three layers work together:

1. **Document** — for exact reference lookups
2. **Vector** — for fuzzy semantic recall
3. **Graph** — for relational reasoning and explainability

### 2.1 From Text to Graph

#### The Pipeline: Raw Text → Knowledge Graph

Building a knowledge graph from support tickets follows four stages:

```
Raw Text          Extracted Facts       Graph Store          Reasoning &
(Tickets, logs,   (Entities &           (Nodes & edges &     Explanation
 or messages)      relationships)        attributes)          (Evidence paths)
```

Each stage transforms the data:
1. **Raw Text** — unstructured tickets, logs, or messages
2. **Extracted Facts** — LLM extracts entities and relationships
3. **Graph Store** — facts stored as nodes, edges, and attributes
4. **Reasoning & Explanation** — agent queries return evidence paths

#### A Simple Support Schema

Starting with a minimal schema keeps things tractable:

```
Nodes:   Customer, Issue, Product
Edges:   reported, affects

Customer --[reported]--> Issue --[affects]--> Product
```

#### Beyond Triplets: Nodes + Edges + Attributes

Real graphs enrich nodes with attributes and edges with metadata:

```
Node — Issue:
  title: "Battery drain"
  severity: "high"
  first_seen: "2024-01-15"

Edge — reported:
  date: "2024-01-16"
  ticket_id: "T-9923"
```

### 2.2 Schema and Ontology Design

#### Ontology as Grammar

An ontology defines the **grammar** of the graph — which relationships are allowed between which node types:

```
Allowed:   Customer  → reported    → Issue
Allowed:   Issue     → affects     → Product
Allowed:   Issue     → resolved_by → Resolution
Forbidden: Product   → reported    → Issue  ✗
```

Without an ontology, agents can insert contradictory or meaningless edges.

#### Entity Extraction with LLM Validation

The LLM extracts candidate entities and relationships from raw text. A validation step checks the extraction against the schema before insertion:

```python
# Pseudocode
raw_ticket = "Customer Alice reports battery drain on Product X"

extracted = llm.extract(raw_ticket)
# → {"entities": [{"type": "Customer", "name": "Alice"},
#                  {"type": "Issue", "title": "battery drain"},
#                  {"type": "Product", "name": "Product X"}],
#    "relationships": [("Alice", "reported", "battery drain"),
#                       ("battery drain", "affects", "Product X")]}

validated = schema.validate(extracted)  # rejects ontology violations
graph.insert(validated)
```

### 2.3 Incremental Graph Construction

#### The Duplicate Challenge

When tickets arrive continuously, naive insertion creates duplicate nodes:

```
Ticket 1: "battery drain"  → creates Issue node #1
Ticket 2: "Battery Drain"  → creates Issue node #2  ← duplicate!
Ticket 3: "batt. drain"    → creates Issue node #3  ← duplicate!
```

Duplicate nodes fragment the graph — related tickets no longer converge on a single issue node, breaking pattern detection.

#### The MERGE / Upsert Solution

Use **MERGE** (upsert) semantics: match on a canonical key and create only if not found.

```cypher
MERGE (i:Issue {title: $normalized_title})
ON CREATE SET i.first_seen = $date
ON MATCH SET  i.count = i.count + 1
```

#### Entity Canonicalization

Before inserting, normalize entity names to a canonical form:

| Raw variant | Canonical form |
|---|---|
| "battery drain" | `battery_drain` |
| "Battery Drain" | `battery_drain` |
| "batt. drain" | `battery_drain` |

Canonicalization is a prerequisite for reliable deduplication.

#### Normalizing Thermal Complaints Across Tickets

The same physical issue appears under different names in different tickets:

- "Laptop overheating" → maps to → **Overheating** (unified node)
- "Thermal shutdown" → maps to → **Overheating** (unified node)
- "Heat warning" → maps to → **Overheating** (unified node)

Two techniques achieve this:
- **Synonym detection** — a lookup table of known equivalents
- **Mapping** — LLM-assisted normalization for novel phrasings

#### Stable Growth Enables Reasoning

| Principle | Details |
|---|---|
| **Incremental Construction** | Normalize entity identifiers; Merge before connecting; Accumulate evidence |
| **Graph Reasoning** | Multi-hop traversal; Causal paths; Pattern detection |

### 2.4 Storage Options: LPG vs RDF

Two dominant paradigms exist for storing knowledge graphs in production:

#### Labeled Property Graphs (LPG)

Examples: **Neo4j**, **TigerGraph**

- **Flexible schema** — nodes and edges carry arbitrary key-value properties
- **Fast traversals** — optimized for graph-native queries
- **Query language** — Cypher (declarative, pattern-matching)
- Best for: operational systems, real-time agent queries, flexible domain models

#### RDF Triple Stores

Examples: **GraphDB**, **Blazegraph**

- **Formal semantics** — built on W3C standards (RDF, OWL, RDFS)
- **Ontology alignment** — supports rich semantic reasoning via inference rules
- **Query language** — SPARQL
- Best for: scientific knowledge, cross-system interoperability, strict ontology compliance

#### Choosing Between Them

| Dimension | LPG | RDF |
|---|---|---|
| Schema flexibility | High | Lower (strict semantics) |
| Query performance | Fast | Moderate |
| Reasoning support | Via plugin | Built-in (OWL/RDFS) |
| Developer ergonomics | Accessible | More specialized |
| Interoperability | Proprietary | Open standards |

### 2.5 Querying the Graph

#### Graphs Return Subgraphs, Not Rows

A graph query doesn't return a flat table — it returns a **subgraph**: a subset of nodes and edges that match the pattern.

This subgraph *is* the evidence the agent uses to form its answer.

#### Three Query Building Blocks

Every useful graph query combines three primitives:

| Building Block | Description | Example |
|---|---|---|
| **Anchor** | Start at a specific node | `MATCH (i:Issue {title: "Battery drain"})` |
| **Traverse** | Follow edges to related nodes | `-[:AFFECTS]->(p:Product)` |
| **Constrain** | Filter by attributes or conditions | `WHERE i.severity = "high"` |

#### Compare Query Syntax: Cypher vs SPARQL

The same question — "Which customers reported a Battery drain issue?" — expressed in each language:

**Cypher** (for property graph databases, e.g. Neo4j):

```cypher
MATCH (c:Customer)-[:REPORTED]->(i:Issue)
WHERE i.title = "Battery drain"
RETURN c, i
```

**SPARQL** (for RDF triple stores):

```sparql
SELECT ?customer WHERE {
  ?customer a ex:Customer .
  ?issue    a ex:Issue .
  ?issue    rdfs:label "Battery drain" .
  ?customer ex:reported ?issue .
}
```

Both express the same traversal: start at Issue, follow `reported` edges in reverse to find Customers.

### 2.6 LLM + Graph Integration

#### The Graph as an LLM Tool Call

The graph is not just data — it is an **evidence source the agent must consult**.

Two integration patterns exist:

**Pattern A — LLM writes the query directly:**

```
Agent (LLM)  ─────────────────────────────────►  Cypher / SPARQL

[LLM understands the schema, and writes queries]
```

**Pattern B — LLM calls a tool, tool writes the query:**

```
Agent (LLM)  ──►  API or function  ──►  Cypher / SPARQL

find_related_issues("Battery Drain")

[LLM doesn't write the query — the tool does]
```

Pattern B (tool-mediated) is more robust in production: the LLM describes *intent*, the tool translates to a validated query.

### 2.7 Hybrid Retrieval

#### Two Strategies for Combining Vectors and Graphs

**Routing strategy** — pick one retrieval tool per query:
- Classify the query type first
- Route to vector store *or* graph store, not both
- Simpler, lower latency

**Fusion strategy** — run both and combine results:
- Query both vector and graph stores in parallel
- Merge and re-rank results before sending to LLM
- Higher recall, more complex

#### What Orion Should Use and Why

| Strategy | When It Fits |
|---|---|
| **Routing** | Graph is reliable for this domain; Repeatable and structured tickets; One evidence source is enough |
| **Fusion** | Graph is still growing; Text is fuzzy / language-heavy; Required evidence in uncertainty |

#### The Right Approach for Orion

How the AI engineer chooses Orion's retrieval policy:

1. **Audit** — Check the data, graph completeness, and language variance
2. **Align** — Identify what is "wrong", user behaviors, and cost of missed cases
3. **Policy** — Based on data nature and business needs

> *In some advanced agents, an AI engineer can choose to orchestrate both based on data and risk.*

### 3.2 Framework Overview

#### Tools for Agentic Knowledge Graphs

Four major options exist, each with a different trade-off profile:

### 3.3 Graph Querying Approaches

#### How Agents Use Graph Queries in Practice

Two patterns for agent-graph interaction:

**Direct query generation** — the LLM understands the schema and writes the query:

```
Agent (LLM)  ─────────────►  Cypher / SPARQL

LLM understands the schema, and writes queries
```

**Tool-mediated** — the LLM calls a function; the function generates and executes the query:

```
Agent (LLM)  ──►  API or function  ──►  Cypher / SPARQL

find_related_issues("Battery Drain")

LLM doesn't write the query
```

#### Tips for Fast Graph Queries

- **Index commonly queried attributes** — e.g., `Issue.title`, `Customer.customer_id`
- **Apply filters to avoid scanning the entire graph** — always anchor on a constrained node first
- **Limit traversal depth for multi-hop patterns** — unbounded hops can be expensive
- **Cache common queries if data is stable** — reduces redundant graph reads

> **Query smart, answer smart: Structured queries help agents retrieve knowledge rapidly and reliably.**

### 3.4 Governance & Maintenance: Making Graphs Production-ready

#### Production Governance Checklist

- **Schema versioning & controlled vocabularies** — track schema changes; maintain canonical term lists
- **Entity resolution & deduplication** — prevent duplicate nodes accumulating over time
- **Data consistency & drift monitoring** — detect when the graph deviates from expected structure
- **Access control & audit logging** — restrict who can read or write which parts of the graph

#### Controlling Access to Sensitive Knowledge

- Only authorized roles can access specific nodes or relationships
- Sensitive nodes are masked, restricted, or excluded from queries
- All access is recorded — who viewed what and when, for accountability
- Schema-based filtering to prevent data leakage

#### Schema Consistency and Evolution

Schemas evolve as the domain grows. Manage evolution carefully:

| Version | Change | Pipeline Impact |
|---|---|---|
| **V1.0** | Schema introduced: `Customer`, `Issue`, `Product` | Ingest v1.0 — schema pinned; stable queries & compatibility |
| **V1.1** | Added `Resolution` linked to `Issue` via `resolved_by` | Post v1.1 — schema validation: only `Issue` can be linked with a `Resolution` |
| **V2.0** | Relationship renaming: `related_to` → `has_variant`, `has_model` | Post v2.0 — renaming watch: monitored for lingering uses of `related_to` |

#### Monitoring Graph Health

A production dashboard should track:

| Metric | What It Signals |
|---|---|
| **Average query latency** | Performance degradation as graph grows |
| **Node growth (by class)** | Unexpected entity type spikes |
| **Orphan node rate** | Extraction failures; disconnected entities |
| **Last index update** | Index staleness |
| **Schema drift alert** | Insertions that violate current schema |

> **Graph governance isn't just operational — it's foundational. It ensures that agents reason over facts that are correct, current, and authorized.**

### 3.5 Demo: Graphs for Multi-hop Reasoning in Agent Workflows

**Scenario:** How an agent uses the graph to trace the impact of a faulty update across multiple nodes.

#### The Multi-hop Query

```cypher
// Find all customers affected by issues triggered by a specific update
MATCH (c:Customer)-[:REPORTED]->(i:Issue)-[:TRIGGERED_BY]->(u:Update {version: "v2.3"})
RETURN c.name, i.title, i.severity
ORDER BY i.severity DESC
```

#### Multi-hop Reasoning Chain

```
Step 1: Anchor on Update v2.3
  → find Issues triggered_by this Update

Step 2: Traverse Issues → Products
  → find which Products are affected

Step 3: Traverse Issues ← Customers
  → find which Customers reported these Issues

Step 4: Check Resolutions
  → are there resolved_by patches already available?
```

#### Agent Response with Evidence Path

```
Answer:
  "Update v2.3 triggered 3 high-severity issues affecting Product X.
   23 customers reported these issues. Patch 1.2 resolves 2 of 3."

Evidence path:
  Update(v2.3) → triggered_by ← Issue(battery_drain) → reported_by ← Customer(×23)
                               → affects → Product(X)
                               → resolved_by → Patch(1.2)
```

The graph doesn't just surface data — it proves the reasoning.
