## Load Env Vars

```python
from dotenv import load_dotenv, find_dotenv
from letta_client import Letta
import os

load_dotenv(find_dotenv())

_letta_key = os.getenv("LETTA_API_KEY")
if not _letta_key:
    raise EnvironmentError(
        "LETTA_API_KEY is not set.\n"
        "Get a free key at https://app.letta.com → Settings → API Keys\n"
        "Then add it to your .env file: LETTA_API_KEY=your-key-here"
    )

client = Letta(api_key=_letta_key)
print("Letta client ready.")
```

## Create Agent
- [Quickstart project](https://app.letta.com/projects/default-project)

```python
# Delete existing agents to stay within the free plan's 3-agent limit
for a in client.agents.list():
    client.agents.delete(a.id)

# Create your agent
messages = [
        {
          "label": "human",
          "value": "Name: Sandy\nOccupation: Software Engineer\nHobbies: Hiking, Cooking, Traveling"
        },
        {
          "label": "persona",
          "value": "You are self-improving superintelligence who act a senior AI engineer."
        }
    ]

agent_state = client.agents.create(
    model="openai/gpt-4o-mini",
    memory_blocks=messages
)

print(f"Agent created with ID: {agent_state.id}")
```

## Running the Agent

```python
def send_message(user_input):
    """Sends a message to the agent and returns the response."""
    return client.agents.messages.create(
        agent_id=agent_state.id,
        input=user_input
    )

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        break

    response = send_message(user_input)

    print(f"\n{Fore.BLUE}🤖 Agent:")
    for message in response.messages:
        print(f"{message.content}{Fore.RESET}")
```
